```python
import io
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset provided
csv_data = """TV Budget ($),Radio Budget ($),Newspaper Budget ($),Sales (units)
230.1,37.8,69.2,22.1
44.5,39.3,45.1,10.4
17.2,45.9,69.3,9.3
151.5,41.3,58.5,18.5
180.8,10.8,58.4,12.9
8.7,48.9,75.0,7.2
57.5,32.8,23.5,11.8
120.2,19.6,11.6,13.2
144.1,16.0,40.3,15.6
111.6,12.6,37.9,12.2"""

# Load dataset into a pandas DataFrame
df = pd.read_csv(io.StringIO(csv_data))

# Define features (X) and target variable (y)
X = df[["TV Budget ($)", "Radio Budget ($)", "Newspaper Budget ($)"]]
y = df["Sales (units)"]

# Initialize and train the multivariate linear regression model
model = LinearRegression()
model.fit(X, y)

# Output model coefficients and intercept
print("Intercept:", model.intercept_)
print("Coefficients (TV, Radio, Newspaper):", model.coef_)

# Example prediction for new data
# e.g., TV = 150, Radio = 25, Newspaper = 40
new_data = pd.DataFrame(
    [[150.0, 25.0, 40.0]],
    columns=["TV Budget ($)", "Radio Budget ($)", "Newspaper Budget ($)"],
)
predicted_sales = model.predict(new_data)
print(f"Predicted Sales (units): {predicted_sales[0]:.2f}")
```

```python
import io
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# Dataset provided
csv_data = """Age,JobRole,MonthlyIncome,JobSatisfaction,YearsAtCompany,Attrition
29,Sales Executive,4800,3,4,1
35,Research Scientist,6000,4,8,0
40,Laboratory Technician,3400,2,6,0
28,Sales Executive,4300,3,3,1
45,Manager,11000,4,15,0
25,Research Scientist,3500,1,2,1
50,Manager,12000,4,20,0
30,Sales Executive,5000,2,5,0
37, Laboratory Technician,3100,2,9,0
26, Research Scientist,4500,3,2,1"""

# Load dataset and clean whitespace from strings and headers
df = pd.read_csv(io.StringIO(csv_data))
df.columns = df.columns.str.strip()
df["JobRole"] = df["JobRole"].str.strip()

# Separate features (X) and target variable (y)
X = df.drop("Attrition", axis=1)
y = df["Attrition"]

# One-hot encode the categorical 'JobRole' feature
X_encoded = pd.get_dummies(X, columns=["JobRole"], drop_first=False)

# Scale features since KNN is distance-sensitive
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_encoded)

# Initialize and train the K-Nearest Neighbors classifier
# (Using n_neighbors=3 due to the small sample size)
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(X_scaled, y)

# Example prediction for a new employee
new_employee = pd.DataFrame(
    [
        {
            "Age": 28,
            "JobRole": "Sales Executive",
            "MonthlyIncome": 4400,
            "JobSatisfaction": 2,
            "YearsAtCompany": 3,
        }
    ]
)

# Align new employee columns with training features and scale
new_employee_encoded = pd.get_dummies(
    new_employee, columns=["JobRole"]
).reindex(columns=X_encoded.columns, fill_value=0)
new_employee_scaled = scaler.transform(new_employee_encoded)

prediction = knn_model.predict(new_employee_scaled)
print("Predicted Attrition (1 = Left, 0 = Stayed):", int(prediction[0]))
```
